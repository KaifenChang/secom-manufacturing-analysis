## Import libraries



```python
# Standard Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning Libraries
from scipy.stats import boxcox
from sklearn.ensemble import RandomForestClassifier
```

## 1. Data processing


```python
# Read Data Files
data_file = "secom/secom.data"
label_file = "secom/secom_labels.data"
df_data = pd.read_csv(data_file, sep="\s+", header=None)
df_label = pd.read_csv(label_file, sep="\s+", header=None, names=["Label", "Timestamp"])

# Merge data and label
df_combined = pd.concat([df_data, df_label], axis=1)

# Assign column names
columns = ["Feature" + str(i) for i in range(df_data.shape[1])]
columns += ["Label", "Timestamp"]
df_combined.columns = columns

# Print summary instead of entire data
print(f"✅ Data loaded successfully: {df_combined.shape[0]} rows, {df_combined.shape[1]} columns.")
print(f"Label distribution:\n{df_combined['Label'].value_counts()}")

# Save to CSV
df_combined.to_csv("secom_combined.csv", index=False)
print("✅ Merged data saved to 'secom_combined.csv'")
```

## 2. Exploratory Data Anaysis

### 2.1 Handling Missing Values
We check for missing values, drop highly missing columns, and impute the remaining ones with the median.


```python
# Load dataset
df_combined = pd.read_csv("secom_combined.csv")

# Compute missing values
missing_values = df_combined.isnull().sum()
missing_percentage = (missing_values / len(df_combined)) * 100
missing_df = pd.DataFrame({"Missing Count": missing_values, "Missing %": missing_percentage})
missing_df = missing_df[missing_df["Missing Count"] > 0]

# Print summary
print(f"🔍 Found {missing_df.shape[0]} features with missing values.")
print("Top 5 missing features:\n", missing_df.head(5))

# Drop columns with >50% missing values
missing_threshold = 50
columns_to_drop = missing_df[missing_df["Missing %"] > missing_threshold].index
df_cleaned = df_combined.drop(columns=columns_to_drop)

print(f"🗑️ Dropped {len(columns_to_drop)} columns with >{missing_threshold}% missing values.")

# Impute remaining missing values with median
numeric_cols = df_cleaned.select_dtypes(include=["number"]).columns
df_cleaned[numeric_cols] = df_cleaned[numeric_cols].fillna(df_cleaned[numeric_cols].median())

# Save cleaned data
df_cleaned.to_csv("secom_cleaned.csv", index=False)
print("✅ Cleaned data saved to 'secom_cleaned.csv'.")
```

### 2.2 Outlier Detection using IQR
We identify and filter outliers based on the Interquartile Range (IQR).


```python
# Compute IQR bounds
Q1 = df_cleaned[numeric_cols].quantile(0.25)
Q3 = df_cleaned[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count outliers per feature
outlier_counts = ((df_cleaned[numeric_cols] < lower_bound) | (df_cleaned[numeric_cols] > upper_bound)).sum()
outlier_counts = outlier_counts.sort_values(ascending=False)

print(f"🔍 Outlier detection complete. {len(outlier_counts[outlier_counts > 0])} features contain outliers.")
print("Top 5 features with most outliers:\n", outlier_counts.head(5))
```

### 2.3 Filtering Outliers (IQR)
We remove extreme outliers while preserving Label = 1 samples.


```python
# Apply IQR filtering while keeping Label = 1 data
filter_condition = ~((df_cleaned[numeric_cols] < lower_bound) | (df_cleaned[numeric_cols] > upper_bound)).any(axis=1)
df_filtered = df_cleaned[(filter_condition) | (df_cleaned["Label"] == -1) | (df_cleaned["Label"] == 1)]

# Save the filtered dataset
df_filtered.to_csv("secom_IQR_filtered.csv", index=False)
print(f"✅ Outliers filtered using IQR. Filtered dataset saved as 'secom_IQR_filtered.csv'.")
```

### 2.4 Visualizing Outliers
We use boxplots and bar charts to understand outlier distributions.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Boxplot of first 10 features (After Outlier Handling)
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_filtered.iloc[:, :10])
plt.xticks(rotation=90)
plt.title("Boxplot of Features After Outlier Filtering")
plt.show()

# Bar Chart: Top 20 Features with Most Outliers
plt.figure(figsize=(12, 6))
outlier_counts.head(20).plot(kind="bar", color="salmon")
plt.title("Number of Outliers per Feature (Top 20)")
plt.xlabel("Features")
plt.ylabel("Outlier Count")
plt.xticks(rotation=90)
plt.show()
```

### 2.5 Capping Extreme Outliers
Instead of removing, we cap extreme values to improve model stability.


```python
# Cap outliers using IQR
df_capped = df_filtered.copy()
df_capped[numeric_cols] = df_capped[numeric_cols].clip(lower=lower_bound, upper=upper_bound, axis=1)

# Ensure Label is preserved
df_capped['Label'] = df_filtered['Label']

# Save capped dataset
df_capped.to_csv("secom_capped.csv", index=False)
print("✅ Outliers capped using IQR and saved to 'secom_capped.csv'.")
```

### 2.6 Skewness Correction using Box-Cox
We identify and correct highly skewed features to improve data normality.


```python
from scipy.stats import boxcox

# Compute skewness
skewness = df_capped[numeric_cols].skew().sort_values(ascending=False)
print("🔍 Top 5 most skewed features before transformation:\n", skewness.head(5))

# Apply Box-Cox Transformation to Highly Skewed Features
df_transformed = df_capped.copy()
highly_skewed_features = skewness[abs(skewness) > 1].index

for feature in highly_skewed_features:
    min_value = df_transformed[feature].min()
    if min_value <= 0:
        df_transformed[feature] += abs(min_value) + 1  # Shift to be positive
    df_transformed[feature], _ = boxcox(df_transformed[feature])

# Save transformed dataset
df_transformed.to_csv("secom_processed.csv", index=False)
print("✅ Skewness transformed and saved to 'secom_processed.csv'.")
```

## 3. Correlation analysis

### 3.1 Compute Correlation Matrix
We calculate the feature correlation matrix to identify relationships between features.


```python
# Calculate correlation matrix for numeric features
correlation_matrix = df_transformed[numeric_cols].corr()

# Save correlation matrix to a CSV for further analysis (optional)
correlation_matrix.to_csv("correlation_matrix.csv")

print("✅ Correlation matrix computed and saved to 'correlation_matrix.csv'.")
```

### 3.2 Visualizing Feature Correlation Matrix
A heatmap helps visualize feature correlations.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot correlation matrix heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=False, cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Features")
plt.tight_layout()
plt.show()

print("✅ Correlation matrix heatmap displayed successfully.")
```

### 3.3 Handling Multicollinearity
We identify and remove highly correlated features (correlation > 0.8) to avoid redundancy.


```python
# Set correlation threshold for feature removal
correlation_threshold = 0.8

# Get the upper triangle of the correlation matrix (to avoid duplicate comparisons)
upper_triangle = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))

# Identify features with correlation greater than the threshold
to_drop = [column for column in upper_triangle.columns if any(upper_triangle[column] > correlation_threshold)]

# Print summary
print(f"🔍 Identified {len(to_drop)} highly correlated features to drop.")
```

### 3.4 Removing Highly Correlated Features
After identifying multicollinear features, we drop them and save the reduced dataset.


```python
# Drop the identified features
df_reduced = df_transformed.drop(columns=to_drop)

# Print remaining feature count
print(f"✅ Removed {len(to_drop)} highly collinear features. Remaining features: {len(df_reduced.columns)}")

# Save reduced dataset
df_reduced.to_csv("secom_reduced.csv", index=False)
print("✅ Reduced dataset saved as 'secom_reduced.csv'.")
```

### 3.5 Compute Feature-Label Correlation
We identify features that have the highest correlation with the target variable (`Label`).


```python
# Compute absolute correlation of features with Label
feature_label_corr = correlation_matrix["Label"].abs().sort_values(ascending=False)

# Print top correlated features with Label
print("🔍 Top 10 features correlated with Label:\n", feature_label_corr.head(10))
```

### 3.6 Visualizing Feature-Label Correlation

This bar chart shows the **top 20 features** most correlated with `Label`.  
Higher correlation suggests **greater predictive power**, while near-zero values indicate **less useful features**.


```python
# Show only the top 20 most correlated features
top_n = 20  
plt.figure(figsize=(12, 6))
feature_label_corr.head(top_n).plot(kind="bar", color="royalblue")

plt.title(f"Feature-Label Correlation (Top {top_n} Features)")
plt.xlabel("Features")
plt.ylabel("Correlation")
plt.xticks(rotation=45)  # Rotate labels for better visibility
plt.show()

print("✅ Feature-Label correlation plot displayed successfully.")
```

## 4. Feature Selection & Comparison

We use **two different methods** to determine feature importance:
1. **Random Forest** – Measures feature importance based on tree splits.
2. **XGBoost** – Uses gradient boosting to evaluate feature contribution.

After computing feature importance, we will:
- **Compare feature rankings** from both models.
- **Visualize importance scores** to identify key features.
- **Select the most relevant features** for further analysis.


```python
print("\n🚀 Calculating feature importance using Random Forest...")

# Prepare the data
X = df_reduced.drop(columns=['Label', 'Timestamp'])
y = df_reduced['Label']

# Convert y to binary classification (0 or 1)
y = (y > 0).astype(int)  # Assigns 1 if y > 0, otherwise 0

# Verify unique values
print("✅ Unique class labels after conversion:", np.unique(y))

# Train Random Forest Classifier
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X, y)

# Compute feature importance
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf.feature_importances_
}).sort_values(by="Importance", ascending=False)

# Save feature importance results
feature_importance.to_csv("random_forest_feature_importance.csv", index=False)
print("✅ Feature importance saved to 'random_forest_feature_importance.csv'.")

# Plot feature importance
plt.figure(figsize=(12, 6))
plt.barh(feature_importance["Feature"].head(20), feature_importance["Importance"].head(20), color="royalblue")
plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Top 20 Feature Importance (Random Forest)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("random_forest_feature_importance.png")
plt.show()

print("✅ Feature importance plot saved as 'random_forest_feature_importance.png'.")
```

```python
print("\n🚀 Calculating feature importance using XGBoost...")

# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from xgboost import XGBClassifier

# Prepare the data
numeric_cols = df_reduced.select_dtypes(include=["number"]).columns.drop('Label')
X = df_reduced[numeric_cols]
y = df_reduced['Label']

# Convert labels from continuous to binary (0 or 1)
y = (y > 0).astype(int)

# Verify unique values
print("✅ Unique class labels after conversion:", np.unique(y))

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

# Train XGBoost Classifier
xgb = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=5,
    min_child_weight=1,
    gamma=0,
    subsample=0.8,
    colsample_bytree=0.8,
    objective='binary:logistic',
    random_state=42,
    scale_pos_weight=1
)

xgb.fit(X_scaled, y)

# Compute feature importance
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': xgb.feature_importances_
}).sort_values(by="Importance", ascending=False)

# Save feature importance results
feature_importance.to_csv("xgboost_feature_importance.csv", index=False)
print("✅ Feature importance saved to 'xgboost_feature_importance.csv'.")

# Plot top 20 features
plt.figure(figsize=(12, 6))
plt.barh(feature_importance["Feature"].head(20), feature_importance["Importance"].head(20), color="salmon")
plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Top 20 Feature Importance (XGBoost)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("xgboost_feature_importance.png")
plt.show()

print("✅ Feature importance plot saved as 'xgboost_feature_importance.png'.")

```

## 5. Comparing Feature Importance: Random Forest vs. XGBoost

We compare **feature importance** extracted from:
1. **Random Forest**
2. **XGBoost**

This section will:
- **Merge** feature importance from both models.
- **Plot a side-by-side bar chart** comparing their feature importance scores.
- **Plot heatmaps** to analyze feature correlation in both models.
- **Select key features** based on their importance in at least one model.


```python
import pandas as pd

# Load Random Forest feature importance
rf_importance = pd.read_csv("random_forest_feature_importance.csv")
rf_importance.rename(columns={"Importance": "RF_Importance"}, inplace=True)

# Load XGBoost feature importance
xgb_importance = pd.read_csv("xgboost_feature_importance.csv")
xgb_importance.rename(columns={"Importance": "XGB_Importance"}, inplace=True)

# Merge on Feature column
feature_comparison = pd.merge(rf_importance, xgb_importance, on="Feature", how="inner")

# Sort by highest importance (average of both models)
feature_comparison["Avg_Importance"] = (
    feature_comparison["RF_Importance"] + feature_comparison["XGB_Importance"]
) / 2
feature_comparison = feature_comparison.sort_values(by="Avg_Importance", ascending=False)

# Save comparison data
feature_comparison.to_csv("feature_importance_comparison.csv", index=False)
print("✅ Feature importance comparison saved as 'feature_importance_comparison.csv'.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Select top 20 features
top_n = 20
top_features = feature_comparison.head(top_n)

# Plot comparison
fig, ax = plt.subplots(figsize=(12, 6))

# Bar width
bar_width = 0.4
indices = np.arange(len(top_features))

# Bar plots
ax.bar(indices - bar_width/2, top_features["RF_Importance"], 
       bar_width, label="Random Forest", color="royalblue")
ax.bar(indices + bar_width/2, top_features["XGB_Importance"], 
       bar_width, label="XGBoost", color="salmon")

# Labels and Titles
ax.set_xticks(indices)
ax.set_xticklabels(top_features["Feature"], rotation=75, ha="right")
ax.set_xlabel("Feature")
ax.set_ylabel("Importance Score")
ax.set_title("Feature Importance Comparison: Random Forest vs XGBoost")
ax.legend()

# Adjust layout and save
plt.tight_layout()
plt.savefig("feature_importance_comparison.png")
plt.show()

print("✅ Feature importance comparison plot saved as 'feature_importance_comparison.png'.")
```

```python
# Define importance threshold
importance_threshold = 0.01

# Keep features that have importance > 0.01 in at least one model
selected_features = feature_comparison[
    (feature_comparison["RF_Importance"] > importance_threshold) |
    (feature_comparison["XGB_Importance"] > importance_threshold)
]["Feature"].tolist()

# Keep only selected features
df_final = df_reduced[selected_features + ["Label"]]

# Save final dataset
df_final.to_csv("secom_final_selected_features.csv", index=False)
print(f"✅ Selected {len(selected_features)} important features.")
print("✅ Final dataset saved as 'secom_final_selected_features.csv'.")
```

```python
import seaborn as sns

# Select top 10 features from each model
top_rf_features = rf_importance.head(10)["Feature"].tolist()
top_xgb_features = xgb_importance.head(10)["Feature"].tolist()

# Ensure they exist in df_reduced
df_rf_top = df_reduced[top_rf_features]
df_xgb_top = df_reduced[top_xgb_features]

# Plot side-by-side heatmaps
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Heatmap for Random Forest
sns.heatmap(df_rf_top.corr(), annot=True, cmap="coolwarm", ax=axes[0])
axes[0].set_title("Feature Correlation Heatmap (Top 10 Features - Random Forest)")

# Heatmap for XGBoost
sns.heatmap(df_xgb_top.corr(), annot=True, cmap="coolwarm", ax=axes[1])
axes[1].set_title("Feature Correlation Heatmap (Top 10 Features - XGBoost)")

# Save and show
plt.tight_layout()
plt.savefig("feature_correlation_heatmaps.png")
plt.show()

print("✅ Feature correlation heatmaps saved as 'feature_correlation_heatmaps.png'.")
```
